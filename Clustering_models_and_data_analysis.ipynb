```python
print("Technology:", len(technology_texts))
print("Sports:", len(sports_texts))
print("Food:", len(food_texts))
print("Total:", len(technology_texts) + len(sports_texts) + len(food_texts))
```

```python
# Create a DataFrame from the three text categories

technology_df = pd.DataFrame({
    "text": technology_texts,
    "category": "Technology"
})

sports_df = pd.DataFrame({
    "text": sports_texts,
    "category": "Sports"
})

food_df = pd.DataFrame({
    "text": food_texts,
    "category": "Food"
})

# Combine all three classes into one dataset
df = pd.concat(
    [technology_df, sports_df, food_df],
    ignore_index=True
)

df.head()
```

```python
# Calculate the number of words in each text

df["word_count"] = df["text"].apply(lambda text: len(text.split()))

df.head()
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["category"]
)

print("Training dataset:", len(train_df))
print("Unseen test dataset:", len(test_df))
```

```python
print("Training data:")
print(train_df["category"].value_counts())

print("\nUnseen test data:")
print(test_df["category"].value_counts())
```

```python
# Select columns needed for Orange

orange_full = df[["text", "category"]]
orange_train = train_df[["text", "category"]]
orange_test = test_df[["text", "category"]]

# Save CSV files

orange_full.to_excel("homework3_full_dataset.xlsx", index=False)
orange_train.to_excel("homework3_training.xlsx", index=False)
orange_test.to_excel("homework3_unseen_test.xlsx", index=False)

print("Files saved successfully.")
```
